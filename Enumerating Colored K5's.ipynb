```python
# These are the matrices that change the order and colors, respectively.

Gleft = MatrixGroup([Permutation([(1,5,8,10,4),(2,6,9,3,7)]).to_matrix(), Permutation([(2,5),(3,6),(4,7),(1,),(8,),(9,),(10,)]).to_matrix()])
Gright = MatrixGroup([Permutation([2,3,4,1]).to_matrix(),Permutation([2,1,3,4]).to_matrix()])
G = cartesian_product([Gleft,Gright]); G.group_generators();
Ggens = list(G.group_generators());
```

```python
#First step: list the 4-colored K5's as 0-1 matrices
I = matrix.identity(4)

# This is basically counting the possible colored graphs which we know is about 816k 

from itertools import product;
L = list(product(range(4),repeat=10));
# 10 elements each being pulled from a set 1,2,3,4.
# This is to represent each shapes. The space corresponds to an edge, and the number corresponds to a color.
colorings = [matrix([I[i] for i in l]) for l in L];
```

```python
#This requires at least one edge per color.

ones = vector([1 for i in range(10)])
colorings2 = [graph for graph in colorings if min(ones*graph) > 0]
```

```python
# These is a broken up list of all color combinations organized by numbers of each color.
# Order matters, Must be from least to greatest for numbers. Otherwise they're empty.
colorings1126 = [graph for graph in colorings2 if (sorted(ones*graph)[0] == 1 and sorted(ones*graph)[1] == 1 and sorted(ones*graph)[2] == 2 and sorted(ones*graph)[3]==6)]
colorings1117 = [graph for graph in colorings2 if (sorted(ones*graph)[0] == 1 and sorted(ones*graph)[1] == 1 and sorted(ones*graph)[2] == 1 and sorted(ones*graph)[3]==7)]
colorings1225 = [graph for graph in colorings2 if (sorted(ones*graph)[0] == 1 and sorted(ones*graph)[1] == 2 and sorted(ones*graph)[2] == 2 and sorted(ones*graph)[3]==5)]
colorings1135 = [graph for graph in colorings2 if (sorted(ones*graph)[0] == 1 and sorted(ones*graph)[1] == 1 and sorted(ones*graph)[2] == 3 and sorted(ones*graph)[3]==5)]
colorings2224 = [graph for graph in colorings2 if (sorted(ones*graph)[0] == 2 and sorted(ones*graph)[1] == 2 and sorted(ones*graph)[2] == 2 and sorted(ones*graph)[3]==4)]
colorings1234 = [graph for graph in colorings2 if (sorted(ones*graph)[0] == 1 and sorted(ones*graph)[1] == 2 and sorted(ones*graph)[2] == 3 and sorted(ones*graph)[3]==4)]
colorings1144 = [graph for graph in colorings2 if (sorted(ones*graph)[0] == 1 and sorted(ones*graph)[1] == 1 and sorted(ones*graph)[2] == 4 and sorted(ones*graph)[3]==4)]
colorings2233 = [graph for graph in colorings2 if (sorted(ones*graph)[0] == 2 and sorted(ones*graph)[1] == 2 and sorted(ones*graph)[2] == 3 and sorted(ones*graph)[3]==3)]
colorings1333 = [graph for graph in colorings2 if (sorted(ones*graph)[0] == 1 and sorted(ones*graph)[1] == 3 and sorted(ones*graph)[2] == 3 and sorted(ones*graph)[3]==3)]
```

```python
# Used to change the numbers below that are dependent on the number of colors
list_colorings = [colorings1117, colorings1126, colorings1144, colorings1135, colorings1333, colorings2224, colorings1225, colorings2233, colorings1234]
numbers_list = [[1,1,1,7], [1,1,2,6], [1,1,4,4], [1,1,3,5], [1,3,3,3], [2,2,2,4], [1,2,2,5], [2,2,3,3], [1,2,3,4]]
```

```python
# This will create unique vectors to represent each graph

for colors in list_colorings:
    x = list_colorings.index(colors)

    # Make sure colors[j] is 2D, not 1D
    if len(colors) <= 0:
        print(f"Skipping color set {x} due to invalid shape")
        continue
    
    # This will permute the edges, and remove redundant graphs
    # Cartisian product
    t = lambda g, x: colors.index(g[0]*colors[x]*g[1])
    H = PermutationGroup(Ggens, action=t, domain=range(len(colors)))
    print(len(H.orbits()))

    reps = []

    # Loop over the orbits and process them
    for o in H.orbits():
        for i in range(len(o)):
            # Ensure we access valid indices
            if (ones * (colors[o[i]]))[0] == numbers_list[x][0] and \
               (ones * (colors[o[i]]))[1] == numbers_list[x][1] and \
               (ones * (colors[o[i]]))[2] == numbers_list[x][2] and \
               (ones * (colors[o[i]]))[3] == numbers_list[x][3]:
                reps.append(o[i])
                break

    print(f"Reps found: {reps}")

    for x in reps:
        y = colors[x]
        graph_as_vector = []
        for j in range(10):
            zeros = vector([0 for i in range(10)])
            zeros[j] = 1
            alpha = zeros * y
            for k in range(4):
                if alpha[k] == 1:
                    graph_as_vector.append(k)
        s = ''.join(str(graph_as_vector))
        print(f"Appending: {graph_as_vector}")
        with open('Canonical_Edge_K5.txt', 'a') as f:
            f.write(s)
            f.write("\n")

```
